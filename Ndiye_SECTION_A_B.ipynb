```python
# write a program that classifies temperatureez into categories using if, elif, and else using the following data
#Below 10 -> cold
#10 to 24 -> Mild
#25 to 34 -> warm
#35 ++ -> Hot

temperature = 125
if temperature < 10:
  print('cold')
elif temperature >= 10 and temperature <= 24:
  print('mild')
elif temperature >= 25 and temperature <= 34:
  print('warm')
else:
  print('hot')
```

```python
humidity_values = [45,60,72,88,35,95,50]
#print each value
for x in humidity_values:
  print(x)

# humidity values above 70
for y in humidity_values:
  if y > 70:
    print(y)

# total humidity
total_humidity = sum(humidity_values)
print(total_humidity)


#average humidity
average_humidity = total_humidity/len(humidity_values)
print(int(average_humidity))
```

###Task B4


```python
def classify_wind_speed(speed):
   if speed >= 40:
       return "Strong Wind"
   elif speed >= 25 and speed <=39:
       return "Windy"
   elif speed >= 10 and speed <=24:
       return "Breezy"
   else:
       return "calm"

print(classify_wind_speed(60))
print(classify_wind_speed(27))
print(classify_wind_speed(16))
print(classify_wind_speed(4))

```
